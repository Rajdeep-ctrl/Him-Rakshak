```python
import pdfplumber
import pandas as pd

pdf_file = "/content/landslide_report_fixed.pdf"

all_rows = []
expected_cols = ["Sl.No.", "Slide_No", "State", "District", "Slide_Name",
                  "NH_SH_Location", "Latitude", "Longitude",
                  "Material_Involved", "Movement_Type", "History"]

with pdfplumber.open(pdf_file) as pdf:
    for i, page in enumerate(pdf.pages):
        tables = page.extract_tables()
        if not tables:
            continue
        table = tables[0]
        for row in table:
            # skip repeated header rows on each page
            if row[0] in (None, "Sl.No.", "") or "Sl.No" in str(row[0]):
                continue
            all_rows.append(row)

df = pd.DataFrame(all_rows, columns=expected_cols[:len(all_rows[0])])
print("Total landslide records extracted:", len(df))
df.head(10)
```

```python
df.to_csv("/content/landslide_raw.csv", index=False)
print("Raw CSV saved:", len(df), "rows")
```
